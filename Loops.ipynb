While Loop :

In while loop, the condition is checked first. If it evaluates to true, the body of the loop is executed otherwise not!

If the loop is entered, the process of [Condition chck and execution] is continued until the condition becomes False.

```python
condition = False
while(condition): # The Block keeps executing until the condition is true
    print(condition)
    # Body of the loop
```

```python
# While Loops :
i = 0
n = int(input("Enter a number: "))
while i <= n:
    print(i)
    i += 1
```

Range function in python

The range() function in python is used to generate a sequence of number:

range(start, stop, step_Size)
### step_size is usually not used with  range()

# An example Demonstrating range() function

```python
for i in range(0,8): # range(8) can also be used
    print(i) # prints 0 to 7
```

### For loop with else

An optional else be used with a for loop if the code is to be executed when the loops exhausts.

```python
l = [3,2,22]
for item in l:
    print(item)
else:
    print("Done") # this is printed when the loop exhausts!
```

## The break statement

'break' is used to come out of the loop when encountered. it instructs the program to exit the loop now!


```python
# example

for i in range(1,6):
    print(i)
    if i==4:
        print(f"Here the number is {i} which is equal to 4 thats why its stopped")
        break
```

## The continue statement

'continue' is used to stop the current iteration of the loop and continue with the next one. it instructs the program to "Skip this iteration"

```python
for i in range(5):
    print("Printing") # this will print 5 times i mean 0 to 4 total 5 and that will skip 5 cause that index of 5 is 6 so that will be skipped!
    if i == 2: # if i is 2, the iteration is skipped.
        continue
    print(i)
```
