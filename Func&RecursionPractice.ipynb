```python
# 01 : WAP using function to find the greatest of three numbers

def great(a,b,c):
    if a > b and a > c:
        return a
    elif b > c and b > c:
        return b
    else:
        return c

a = int(input("Enter a number: "))
b = int(input("Enter another number: "))
c = int(input("Enter another number: "))

print(f"The greatest number is: {great(a,b,c)}")
```

```python
# 01 WAP using function to convert Celsius to Fahrenheit
# c /5 = f- 32/9

def cTof(x):
    a = ((x/5*3)+32)
    return a

a = int(input("Enter a number: "))
print(f"Celsius ({a}) to Fahrenheit is : {round(cTof(a),2)}") # here we used round(number, ndigits) here
# number: the required value  to be rounded. this can be an int or a float and
# ndigits: An optional int that specifies the number of decimal places to round to. the default value is 0
# if we use like default : print(round(16.2)) output will be 16 and .2 will be out {
# rounding to the nearest int(ndigits omitted or 0 ) : The function returns the nearest whole number. the return type is an int!
# }
# in print(round(1.11212,2)) output will be 1.11 {
# rounding to a specific number of decimal places (ndigits is positive ): the function returns a float rounded to that many digits after decimal point.
# }
```

```python
# 03 Write a recursive function to calculate the sum of first n natural numbers.
def rFunc(n):
    if n == 0:
        return 0
    return n + rFunc(n-1)

a = int(input("Enter a number: "))
print(f"Recursive function is {rFunc(a)}")
```
