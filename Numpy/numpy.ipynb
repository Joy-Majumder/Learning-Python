### record array or structured array

- We can specify a type and, optionally, a name on a per-column basis. This makes sorting and filtering even more powerful, and it can feel similar to working with data in

### By default, NumPy arrays (ndarray) are homogeneous, meaning all elements must be of the same data type.
- Homogeneous arrays allow:
    1. Vectorized operations
    2. Efficient memory layout (contiguous blocks)
    3. Fast computations using C under the hood
- NumPy is designed for arrays of fixed-size elements — typically numerical data type (integers, floats), but also supports strings, booleans, dates, and custom structured types

- All elements in a NumPy array share the same data type ( dtype ).
- The data is stored in a contiguous block of memory for efficiency.

It can support heterogeneous types using object arrays or structured types, but these lose most performance benefits.

### create a record array from a(flat) list of arrays

`````python
x1 = [1,2,3]
x2 = [2.0,2.0,6.0]
x3 = ['a','b','c']
`````

`````python
a1 =list(zip(x1,x2,x3))
a2 = np.array(a1) # if someone create arr from plain data using this method this is called : Homogeneous array
a2 
# here size of a2 and r is totally diffrent and a2 side is massively bigger than r even through they hold the exact same value/ chars
# for example : 
# In r:
# we gave it strict blueprints: ('a','i4'), ('b','f4'), ('c','U1')
# Slot A (i4): takes 4 Bytes
# Slot B (f4): takes 4 Bytes
# Slot C (U1): takes 4 Bytes * Total RAM used for this row: 12 Bytes.
#  IN a2 : 
# slot a -> 16 bytes
#  slot B -> 16 bytes
#  slot c -> 16 bytes  # total is 48bytes

r = np.array(list(zip(x1,x2,x3)), dtype=[('a','i4'),('b','f4'),('c','U1')]) # and this one called Structured Record array
# r = np.array(list(zip(x1,x2,x3)), dtype=object) # and this one called Structured Record array
r # and here 'a','b', 'c' is the nickname or keys for the arrays 1,2,3
`````

### Broadcasting 

Broadcasting Rules
1. Compare shapes from the last dimension to the first.
    - If one shape has fewer dimensions, leading 1s are added automatically to the
      smaller shape.
2. Dimensions are compatible if:
    - They are equal, or
    - One of them is 1
3. Resulting shape is the maximum in each dimension after broadcasting.

`````python
# simple Example (1D + scalar)
arr01 = np.array([1,2,3])
print(arr01+5) # here broadcasts 5 to [6,7,8]
`````

`````python
arr02 = np.array([[1,2,3],
                  [4,6,2]])
arr01 = np.array([1,2,3])
chu = arr02 + arr01
print(chu)
`````

`````python
# Broadcasting Fails When Shapes are Incompatible
pl = np.array([[1,2,4],
               [4,5,6]])
pl2 = np.array([[10,20],
               [40,50]])
`````

`````python
pl3=(pl + pl2)
print(pl3) # here we can see that fails to broadcasting so yea in these types broadcasting fails
`````

#### Broadcasting Internals Explained with Examples

1. Broadcasting in NumPy is implemented in C under the hood for speed.
2. Align shapes: (2,3) and (1,3)
3. Strides for B (row stride = O) → Reuse the same row twice
4. Compute element-wise sum in a new (2,3) array - using Vectorization

#### Advantages of Broadcasting
1. Memory efficient (no actual data replication).
2. Faster computations using vectorization instead of loops.

### 5. Vectorization

- Vectorization means performing operations on entire arrays at once instead of iterating
element by element in Python.
- NumPy operations are implemented in C, which avoids Python-level loops and
interpreter overhead.
