# File I/O
The random access memory is volatile, and all its contents are lost once a program terminates in order to persist the data forever, thats why we use files.<br>
A file is data stored in a storage device. A python program can talk to the files by reading content from it and writing content to it. <br>
RAM = volatile <br>
HDD = non-volatile

### Type of files

there are 2 types of files: <br>
1. Text files <br>
2. Binary files <br>
python has a lot of functions for reading, updating and deleting files.

### Opening a file
python has an open() function for opening files. It takes 2 parameters: filename and mode. <br>


```python
# open("filename", "mode of opening(read mode by default)")
f = open("main.py","r")
# text = f.read() # print its contents
# print(text)
# f.close() # by using this function we can close file..

# other methods to read a file.
# we can also use f.readline() function to read one full line at a time.
text = f.readline() # by using this we can read line by line lol
print(text)
f.close() # whenever we open file we should always close the file.
```

```python
# Write files
f = open("main1.py","w")
f.write("print(\"hello world\")") # write a str to the file
# f.close()
f = open("main1.py","w")
print(f)
f.close()
```
