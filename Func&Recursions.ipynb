## Default parameter value

we can have a value as default as default argument in a function.<br>
If we specify name = "Joy" in the line containing def, this value is used when no argument is passed.

```python
# def fre(name = "Joy"):
def fre(name, ending="Thanks"):
    print(f"Hello {name} {ending}" )

# fre() # name will be "Joy" in the function body (default)
fre("Code") # name will be "Code" in function body (passed)
fre("Joy", "Thanku") #  if i pass the ending value here "Thanku" then it will return thanku and default thanks will not be print lol also if i just pass the value of name without ending then it will print default ending
fre("jazz") # here we can see that here returns thanks cause i didnt pass ending value just passed name
```

recursion is a function which call itself.<br>
it is used to directly use a mathematical formula as function.

```python
# example here :

def factorial(n):
    if n == 1 or n== 0 :
        return 1
    return n * factorial(n-1)

n = int(input("Enter a number here : "))
print(f"Factorial of {n} will be : {factorial(n)}")
```

### Diagrams

```python
def fact_diagram(n, chain=""):
    if n == 1:
        print(chain + "1")
        return 1
    print(chain + f"{n} x factorial({n-1})")
    return n * fact_diagram(n-1, chain+ f"{n} x ")

print("Factorial(5)")
result = fact_diagram(5)
print("\n")
print(f"Factorial of 5 is : {result}")
```

The programmer needs to be extremely careful while working with recursion to ensure that the function doesnt infinitely keep calling itself.<br>Recursion is sometimes the most direct way to code on algorithms.
