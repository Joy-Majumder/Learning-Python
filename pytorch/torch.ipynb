## creating a Tensor
- also this is kinda similar with the numpy

```python
# using empty 
torch.empty(2,3) # so it is basically making a 2d tensor matrix; also it doesnt assign the value it is actually doesnt assign the value so it is actually assigning the already valued
```

```python
type(torch.empty(2,3)) # by using this we can actually claim that we have created a thing which is a tensor so yea 
```

```python
# Using the Zeros
torch.zeros(2,3) # so it is as like numpy zeros so all tensors are assigned zeros
```

```python
# Using Ones
torch.ones(2,3) # same as zero
```

```python
# Using Rand -> rand means random 
torch.rand(2,3) # so it basically creates the values between zero to ones
```

```python
# use the seed
torch.rand(2,3)
```

```python
# manual seed -> so suppose if we want to reproduce the same value of the rand val sow we can use these. 
torch.manual_seed(100)
torch.rand(2,3)
```

```python
# manual seed 
torch.manual_seed(100)
torch.rand(2,3) # So by using this if a value of rand provides me a vlaue and i wanwt that my fd should see the same value so yea we could use seeed that time 
```

```python
# Using Tensor 
torch.Tensor([[1,2,4],[1,2,3]]) # by using this it will make tensor of my providing values so yea 
```

```python
# other ways
# arange
print("Arrange using : ", torch.arange(0,10)) # range it is and in the last we can provide the step (0,10,2)

# using linspace
print("Using LinSpace : ", torch.linspace(0,10,10))

# using the eye
print("Using eye : ", torch.eye(5))

# using the full 
print("Using full : ", torch.full((3,3),5))
```
