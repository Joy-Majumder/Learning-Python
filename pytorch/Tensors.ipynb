## Tensor Operations on GPU


```python
torch.cuda.is_available()
```

```python
device = torch.device('cuda')
```

```python
# Creating a new tensor on gpu
torch.rand((2,3),device=device)
```

```python
# moving an existing tensor on gpu
print(a) # this is based on cpu
b = a.to(device) # this is moved to gpu 
b
```

```python
b + 5 # now we are performing this on gpu
```

```python
import torch
import time

size = 10000

matrixOnCpu1 = torch.randn(size, size)
matrixOnCpu2 = torch.randn(size, size)

startTime = time.time()
resultCpu = torch.matmul(matrixOnCpu1, matrixOnCpu2)
cpuTime = time.time() - startTime

print(f"Time on cpu : {cpuTime:.2f} seconds ")

matrixOnGpu1 = matrixOnCpu1.to('cuda')
matrixOnGpu2 = matrixOnCpu2.to('cuda')

# time measure on gopu
startTime = time.time()
resultGpu = torch.matmul(matrixOnGpu1, matrixOnGpu2)
torch.cuda.synchronize() # by using this we are ensuring that all operations on gpu done perfectly 
gpuTime = time.time() - startTime

print(f"Time on Gpu : {gpuTime:.4f} seconds\n")

# comparing the results 
print("SpeedUp (cpu time/ gpu time) : ", cpuTime/gpuTime)
```
