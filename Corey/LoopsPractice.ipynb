```python
# 04
n = int(input("Enter a number here : "))
for i in range(2,n):
    if(n%i==0):
        print("Its not prime number")
        break
else:
    print("Its prime number")
```

```python
# 05
n = int(input("Enter a number here : "))
i = 1
sum = 0
while(i<=n):
    sum  += i
    i+=1
print(sum)
```

```python
# 06
# 5! = 1*2*3*4*5 = 120
n = int(input("Enter a number here : "))
product = 1
for i in range(1,n+1):
    product *= i
print(f"The factorial of {n} is {product}")
```
