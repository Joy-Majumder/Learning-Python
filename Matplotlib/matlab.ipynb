```python
# Matplotlib Pyplot
# 
import matplotlib.pyplot as plt

# 1. We just start plotting! Pyplot auto-creates the Figure and Axes in the background.
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 45]

plt.plot(x, y, color='crimson', marker='o', linewidth=2, label='Speed (mph)')

# 2. These commands automatically attach to the active plot we just made above
plt.title("Pyplot State Machine in Action", fontsize=14, fontweight='bold')
plt.xlabel("Time (Seconds)")
plt.ylabel("Speed")
plt.grid(True, linestyle='--')
plt.legend()

# 3. Render the final result to your screen and clear the state machine!
plt.show()
```
